# Segmentation — deferred roadmap

**Intentionally not implemented.** **Decision it would serve:** a different service, schedule or capacity choice per segment. A segmentation that changes no decision is not built.

**Evidence it needs:** validated states first, then recency, frequency, trend and consistency with tenure and coverage, compared against what operations already does.

**What the monthly data already shows:** month-to-month changes (39.8% fewer, 28.5% equal, 31.7% more visits) and population patterns by month of run. Clustering these monthly aggregates would produce groups without an operational meaning, so it is not attempted.

## Gate before any code

1. Event-level check-ins with verified capture, time zone, closures, pauses and entitlement.
2. A written operational decision that would change, reviewed with the people who make it.
3. Thresholds calibrated on distributions and checked for stability over time, with an explicit 'unknown' outcome when evidence is missing.
4. Contractual outcomes kept in the Retention module; behavior features never borrow its labels.

Neither the Berkeley monthly panel (counts per month, 1997–2001) nor the Mindweave synthetic sample meets gate 1. See `docs/behavioral_state_definition.md` and `handoff/module_spec.md`.
