# Member states — deferred roadmap

**Intentionally not implemented.** **Decision it would serve:** who needs onboarding support, routine service, a barrier review or a capture check, instead of one message for everyone.

**Candidate states (hypotheses, not a taxonomy):** New, Highly engaged, Stable, Declining, Low engagement, Dormant, plus Unknown when capture is unverified.

**What the monthly data already shows:** 30.4% of eligible regular months record no visit and 16.1% of complete three-month windows have none (notebook 07, site). These describe volume; they do not assign anyone a state, because a monthly zero cannot separate absence from missing capture.

## Gate before any code

1. Event-level check-ins with verified capture, time zone, closures, pauses and entitlement.
2. A written operational decision that would change, reviewed with the people who make it.
3. Thresholds calibrated on distributions and checked for stability over time, with an explicit 'unknown' outcome when evidence is missing.
4. Contractual outcomes kept in the Retention module; behavior features never borrow its labels.

Neither the Berkeley monthly panel (counts per month, 1997–2001) nor the Mindweave synthetic sample meets gate 1. See `docs/behavioral_state_definition.md` and `handoff/module_spec.md`.
