# Behavioral risk — deferred roadmap

**Intentionally not implemented.** **Decision it would serve:** when a person's attendance pattern justifies an optional, reviewed conversation about barriers.

**Evidence it needs:** validated states, within-member baselines and independently adjudicated outcomes. Behavioral risk must never be presented as churn probability.

**What exists today:** the separate Retention module holds a bounded historical target and benchmark on the same Berkeley population. This lab shares keys with it (81,927 rows matched) but imports none of its outcomes or scores, so attendance features cannot leak into a churn label.

## Gate before any code

1. Event-level check-ins with verified capture, time zone, closures, pauses and entitlement.
2. A written operational decision that would change, reviewed with the people who make it.
3. Thresholds calibrated on distributions and checked for stability over time, with an explicit 'unknown' outcome when evidence is missing.
4. Contractual outcomes kept in the Retention module; behavior features never borrow its labels.

Neither the Berkeley monthly panel (counts per month, 1997–2001) nor the Mindweave synthetic sample meets gate 1. See `docs/behavioral_state_definition.md` and `handoff/module_spec.md`.
